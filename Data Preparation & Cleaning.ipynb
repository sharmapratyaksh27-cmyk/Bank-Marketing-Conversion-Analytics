```python
import pandas as pd
```

```python
df = pd.read_csv("C:\\Users\\karti\\Downloads\\Bank_Marketing_Dataset.csv")

print("Dataset Shape:", df.shape) 
print(df.head())
```

```python
column_mapping = {
    'V1': 'age',
    'V2': 'job',
    'V3': 'marital',
    'V4': 'education',
    'V5': 'default',
    'V6': 'balance',
    'V7': 'housing',
    'V8': 'loan',
    'V9': 'contact',
    'V10': 'day',
    'V11': 'month',
    'V12': 'duration',
    'V13': 'campaign',
    'V14': 'pdays',
    'V15': 'previous',
    'V16': 'poutcome',
    'Class': 'subscribed'
}

df.rename(columns=column_mapping, inplace=True)
```

```python
df['subscribed'] = df['subscribed'].map({1: 'no', 2: 'yes'})

print(df['subscribed'].value_counts())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
duplicates = df.duplicated().sum()
print(f"\nNumber of duplicate rows: {duplicates}")
```

```python
if duplicates > 0:
    df = df.drop_duplicates()
```

```python
df.to_csv('Bank_Marketing_Cleaned.csv', index=False)
print("Data cleaning complete. Cleaned file saved!")
```
